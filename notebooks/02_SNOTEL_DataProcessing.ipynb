```python
# Download data from SNOTEL Report Generator 2.0
# https://wcc.sc.egov.usda.gov/reportGenerator/

# CSV processing to get peak SWE/DOY report
import os
os.chdir('../data')

import pandas as pd

df = pd.read_csv(r'SNOTEL/SNOTEL_swe_20042024.csv')
df.columns = ['date', 'swe_in']  # rename columns
df['date'] = pd.to_datetime(df['date'])
df['swe_m'] = df['swe_in'] * 0.0254  # convert inches to meters
df = df.sort_values('date').reset_index(drop=True)

print(df.head())
print(f'{len(df)} total records')

df.to_csv('SNOTEL/SNOTEL_swe_20042024_cleaned.csv')
```

```python
# build a list to collect results, one year at a time
rows = []

# Loop through each unique year in df
for year in df['date'].dt.year.unique():
    
    # Filter df down to only rows from this year
    year_df = df[df['date'].dt.year == year]

    # Find the index (row number) of the maximum SWE in that filtered data
    peak_idx = year_df['swe_m'].idxmax()

    # Use that index to grab the entire row
    peak_row = year_df.loc[peak_idx]

    # Append the values we care about to our list
    rows.append({
        'year': peak_row['date'].year,
        'date': peak_row['date'],
        'peak_swe_m': peak_row['swe_m']
    })

# Convert the list of dictionaries into a DataFrame
df2 = pd.DataFrame(rows)

df2.to_csv('SNOTEL/SNOTEL_20042024_peakswe_m.csv')

print(df2)
```
